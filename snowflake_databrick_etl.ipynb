## Databricks ↔ Snowflake Connectivity
> Databricks uses the Snowflake Spark Connector to read and write data directly to Snowflake tables. The connector supports query pushdown, so filtering and aggregation can execute in Snowflake for better performance.

For SNOWFLAKE 
- CREATE DATABASE SALES_DB;

- SHOW DATABASES;

- USE DATABASE SALES_DB;

- CREATE SCHEMA SALES_DB_SC;
